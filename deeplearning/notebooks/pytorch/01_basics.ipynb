### Inplace Operations


```python
m = torch.rand(2,3)
n = torch.rand(2,3)

print(m)
print(n)
```

```python
m+n
```

```python
m.add(n)  # This is another way to add two tensors together
```

```python
m
```

```python
n
```

```python
torch.relu(m)   # Relu is a activation function which is used in deep learning to convert all the negative values to zero and keep the positive values as it is . This function will create a seperate tensor with the relu values but it will not change the original tensoer m  so if we want to change the original tensor m then we can use inplace operation by adding _ at the end of the function name 
```

```python
m.relu_() # This _ function represents an in-place operation . This will change the original tensor m and convert all the negative values to zero and keep the positive values as it is . 
```

### Copying a Tensor



```python
a = torch.rand(2,3)
a
```

```python
b = a # We are making a copy of a in b but it is not a deep copy it is a shallow copy because if we change the value of a then the value of b will also change because they are pointing to the same memory location 
```

```python
a[0][0]= 0
a
```

```python
b
```

```python
# so to prevent this type of problem we can use the clone function to make a deep copy of the tensor 
b = a.clone()
a[0][0]= 1
a
```
