### Tensor Operations on GPU

```python
torch.cuda.is_available()  # This function is used to check if the cuda is available or not in our system . If it returns true then we can use the cuda to perform operations on GPU which will be faster than CPU . If it returns false then we can only use CPU for our operations .
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')  # This line of code is used to set the device to cuda if it is available otherwise it will set the device to cpu . This is a common practice in deep learning to make the code compatible with both CPU and GPU .
print(device)
```

```python
# Creating new tensor on the device 
torch.rand(2,3).to(device)
```

```python
# Moving an existing tensor to the device
a
```

```python
b = a.to(device)
b
```

```python
# Comparision between CPU and Cuda Speed
import time


# Define the size of the tensor
size = 1000 # Large size for better comparison

# Create random matrices on CPU

matrix_cpu1 = torch.rand(size,size)
matrix_cpu2 = torch.rand(size,size)

# Measure time on CPU
start_time_cpu = time.time()
result_cpu = torch.matmul(matrix_cpu1, matrix_cpu2)     # Matrix multiplication on cpu
cpu_time = time.time() - start_time_cpu
print(f"CPU Time: {cpu_time:.4f} seconds")

# Move Matrices to GPU
matrix_gpu1 = matrix_cpu1.to(device)
matrix_gpu2 = matrix_cpu2.to(device)

# Measure time on GPU
start_time_gpu = time.time()
result_gpu = torch.matmul(matrix_gpu1, matrix_gpu2)     # Matrix multiplication on gpu
torch.cuda.synchronize()  # Wait for GPU to finish
gpu_time = time.time() - start_time_gpu

print(f"GPU Time: {gpu_time:.4f} seconds")

# compare the results
print(f"Results are close: {torch.allclose(result_cpu, result_gpu.cpu())}")

```
