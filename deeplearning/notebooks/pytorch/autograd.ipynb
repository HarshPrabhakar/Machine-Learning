### another perceptron  Without Autograd


```python
import torch
```

```python
# Inputs
x = torch.tensor(6.7)   # input feature
y = torch.tensor(0.0)   # true label 

w  = torch.tensor(1.0)     # weight (parameter to be learned)
b = torch.tensor(0.0)     # bias (parameter to be learned)
```

```python
# Binary Cross-Entropy Loss
def binary_cross_entropy_loss(prediction, target):
    epsilon = 1e-7  # small constant to avoid log(0)
    prediction = torch.clamp(prediction, epsilon, 1 - epsilon)  # clamp predictions to avoid log(0)
    loss = - (target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
    return loss
```

```python
# Forward pass
z = w*x +b # Weighted sum (linear transformation)
y_pred = torch.sigmoid(z)  # Sigmoid activation to get predicted probability

# Compute binary cross entropy loss
loss = binary_cross_entropy_loss(y_pred, y)
loss
```

```python
# Derivative of loss with respect to w and b
# 1st dl/dy_pred : loss with respect to predicted probability
dloss_dy_pred = (y_pred -y )/(y_pred * (1 - y_pred))  # derivative of binary cross-entropy loss with respect to predicted probability   

# 2nd dy_pred/dz : predicted probability with respect to linear output
dy_pred_dz = y_pred * (1 - y_pred)  # derivative of sigmoid function with respect to its input

# 3rd dz/dw : linear output with respect to weight
dz_dw = x  # derivative of linear output with respect to weight
dz_db = 1  # derivative of linear output with respect to bias 

dl_dy = dloss_dy_pred * dy_pred_dz  * dz_dw # chain rule to get derivative of loss with respect to linear output
dl_db = dloss_dy_pred * dy_pred_dz  * dz_db # chain rule to get derivative of loss with respect to bias
```

```python
print("Gradient of loss with respect to weight w:", dl_dy)
print("Gradient of loss with respect to bias b:", dl_db)    
```

### Another perceptron with AutoGrad


```python
x = torch.tensor(6.7)   # input feature
y = torch.tensor(0.0)   # true label
```

```python
w = torch.tensor(1.0, requires_grad=True)     # weight (parameter to be learned
b = torch.tensor(0.0, requires_grad=True)     # bias (parameter to be learned)
```

```python
w
```

```python
b
```

```python
z = w*x +b # Weighted sum (linear transformation)
y_pred = torch.sigmoid(z)  # Sigmoid activation to get predicted probability
y_pred

```

```python
loss = binary_cross_entropy_loss(y_pred, y)
```

```python
loss
```

```python
loss.backward()  # compute gradients of loss with respect to w and b
```

```python
print(w.grad)
print(b.grad)
```
