```python
import torch
```

```python
print(torch.__version__)
```

```python
#  This is a simple method to create a 1D Tensor from pytorch

x = torch.tensor([1,2,3,4,5])

print(x)
```

```python
# Creating a multi dimensional Tensor in torch

y = torch.tensor([[1,2],[3,4]])
print(y)
```

```python
# Another way to create a tensor in zeros

z = torch.zeros(3,4)  # 3 belongs to rows and 4 belongs to columns
print(z)
```

```python
# A tensor filled with values of ones 1
x = torch.ones(2,4)
x
```

```python
# Another way to create tensor by rand function
r = torch.rand(3,5)  # So this function will create list of random decimal numbers 
r
```
