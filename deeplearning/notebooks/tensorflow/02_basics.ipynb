```python
import torch
```

```python
x = torch.tensor([1.,2.,3.,4.])

# step 2 
 
y = torch.tensor([2.,4.,6.,8.])

# wrong guesses for w and b
w = torch.tensor(0.0)
b = torch.tensor(0.0)

# step 4 
y_pred = w * x+b

loss = ((y_pred- y)**2).mean()

print("predicted values :",y_pred)

print("Actual values :",y)

print("Loss :",loss)
```
