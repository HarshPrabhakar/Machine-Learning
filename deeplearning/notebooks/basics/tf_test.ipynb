```python
# Importing tensorflow

import tensorflow as tf
```

```python
# Creating Tensors : You can create tensors using TensorFlow’s tf.Tensor() or its various helper functions, such as tf.constant(), tf.Variable() or tf.zeros():

scalar_tensor = tf.constant(5)
vector_tensor = tf.constant([1, 2, 3, 4])

matrix_tensor = tf.constant([[1, 2], [3, 4]])
tensor_3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
zeros_tensor = tf.zeros([3, 3])
ones_tensor = tf.ones([2, 2])

print(scalar_tensor,vector_tensor,matrix_tensor,tensor_3d,zeros_tensor,ones_tensor)
```

```python
# Teansor Operation : TensorFlow supports various operations that can be performed on tensors, such as element-wise operations, matrix multiplication, reshaping and more.

matrix_tensor = tf.constant([[1, 2], [3, 4]])

ones_tensor = tf.ones_like(matrix_tensor)
tensor_add = tf.add(matrix_tensor, ones_tensor)
matrix_mult = tf.matmul(matrix_tensor, matrix_tensor)
reshaped_tensor = tf.reshape(matrix_tensor, [4, 1])

transpose_tensor = tf.transpose(matrix_tensor)

print(matrix_tensor,ones_tensor,tensor_add,matrix_mult,reshaped_tensor,transpose_tensor)
```

```python
# Accessing Elements in Tensor : You can access specific elements within a tensor using indices. Similar to how you access elements in Python lists or NumPy arrays, TensorFlow provides slicing and indexing operations.

vector_tensor = tf.constant([1, 2, 3, 4])
first_element = vector_tensor[0]
matrix_tensor = tf.constant([[1, 2], [3, 4]])

matrix_slice = matrix_tensor[:2, :]

print(vector_tensor,first_element,matrix_tensor,matrix_slice)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

X_train = tf.random.normal([100, 10])
y_train = tf.random.normal([100, 1])

model = Sequential([
    Dense(64, activation='relu', input_shape=(10,)),
    Dense(32, activation='relu'),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')

model.fit(X_train, y_train, epochs=10)
```
