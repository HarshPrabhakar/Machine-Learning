### Train - Test - Split

```python
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size = 0.2)
```

### Numpy arrays to Pytorch tensors

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

## Defining the Model

```python
class MysimpleNN():
    def __init__(self,input_size):
        self.weights = torch.rand(input_size,1,dtype= torch.float64,requires_grad = True)
        self.bias = torch.zeros(1,dtype= torch.float64,requires_grad = True)

    def forward(self,X):
        z = torch.matmul(X,self.weights) + self.bias
        y_pred = torch.sigmoid(z)
        return y_pred

    def loss_function(self,y_pred,y):
        epsilon = 1e-7
        y_pred = torch.clamp(y_pred,epsilon,1-epsilon)
        # Calculate loss
        loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
        return loss
```

### Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

### Training Pipeline


```python
# Create  Model
model = MysimpleNN(input_size = X_train_tensor.shape[1])
model.weights
# define loop
for epoch in range(epochs):
    # Forward pass
    y_pred = model.forward(X_train_tensor)
    

    # loss calculation
    loss = model.loss_function(y_pred,y_train_tensor)

    # backward pass
    loss.backward()

    # parameter update
    with torch.no_grad():

        model.weights.data -= learning_rate * model.weights.grad
        model.bias -= learning_rate * model.bias.grad

    # zero gradients

    model.weights.grad.zero_()
    model.bias.grad.zero_()

    # print loss in each epoch
    print(f'Epoch {epoch+1}, Loss: {loss.item()}')
```

### Model Evaluation

```python
# evaluate the model on test data
with torch.no_grad():
    y_pred = model.forward(X_test_tensor)
    y_pred_labels = (y_pred > 0.9).float()
    accuracy = (y_pred_labels.squeeze() == y_test_tensor).float().mean()
print(f"Test Accuracy: {accuracy}")
```

### Building an Neural Network with Torch.nn Modeule


```python
import torch.nn as nn
```

```python
#  Create Model class using nn.Module

class model(nn.Module):

    def __init__(self,num_features):         # constructor to initialize the layers of the model and num_features is the number of features in the dataset .
        super().__init__()                   # invoke the constructor of the parent class nn.Module
        self.linear = nn.Linear(num_features,1)  # Linear layer takes all the features and produces a single output for binary classification
        self.sigmoid = nn.Sigmoid()             # Sigmoid activation function to convert the output of the linear layer into a probability between 0 and 1
    
    def forward(self,features):
        out = self.linear(features)              # pass the input feature through the linear layer to get the raw output (logits)
        out = self.sigmoid(out)                 # pass the logits through the sigmoid activation function to get the predicted probabilities
        return out

    
```

```python
# create dataset
features = torch.rand(10,5)
# Create Model
model = model(features.shape[1])
# call model for forward pass
model(features)

```
