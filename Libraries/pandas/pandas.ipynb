```python
import pandas as pd
import numpy as np 
```

```python
df = pd.read_csv("C:\ML_Scikit\Datasets\data-.csv")
print(df.head())
```

```python
### This will help you to understand about the data its data type and its properties. Like how many column it contains , how manu null values each row contains and which d_type it stores and how many bytes of data it stores.

print(df.info())
```

```python
## Removing the Null Values Mainly the data that arrives has many loop holes many null values so we have to fix it if we don't the computer will not be able to understand it and will generate an error.

print(df.isnull())   # this will print the Null values from each column 

print(df.isnull().sum())    # Now this will comnine all the column and return values in either 0 or 1 if it is zero that means that column is not having any null values and if it returns 1 it means it has some null values that we have to fix .

df = df.fillna(0)       # if you found any null values then this function will fill all the null values with a constant number of 0 which will help the programmer to execute code efficiently
```

```python
ages = df[df['age'] > 25]  # Targetting a particular column for modification.
print(ages)
```

```python
#  Adding and Removing Columns: You can create new columns based on existing ones or delete unwanted columns from the DataFrame.

df['total'] = df['a'] + df['b']
df.head()
```

```python
#  Grouping Data (GroupBy): Grouping allows you to organize data into categories and compute values for each group for example, sums, counts or averages.

res = df.groupby('category')['sales'].sum()
print(res)
```
